```python
import pandas as pd
import numpy as np
```

```python
d={
    "Name":["Ram","Rohan","Denny"],
    "Marks":[70,None,89],
    "Salary":[25000,30000,None],
    "Branch":["CSE",None,"AIML"]
}
df=pd.DataFrame(d)
print(df)
```

```python
df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
df["Salary"]=df["Salary"].fillna(df["Salary"].median())

print(df)
```

```python
print("Before forward fill")
print(df)
```

```python
df["Branch"]=df["Branch"].ffill()
print("After foward fill")
```

```python
print(df)
```

```python
print("Representing True or False in the form 0  and 1")
encoded_df=pd.get_dummies(df,columns=["Branch"],dtype=int)
print(encoded_df)
```
