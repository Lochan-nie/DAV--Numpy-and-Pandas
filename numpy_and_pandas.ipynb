```python
import pandas as pd
import numpy as np
a=np.array([10,20,30,40,50])
print("Array :",a)
print("Mean :",a.mean())
print("Reshaped 2D :\n",a.reshape(5,1))
print("\nSlice (index:(1:3)) :",a[1:4])


d={
    "Name":["Ram","Rohan","Denny"],
    "Marks":[70,85,89],
    "Branch":["CSE","ECE","AIML"]
}
da=pd.DataFrame(d)
print("\n\n")
print(da)
print("First two rows :\n",da.iloc[0:2])
print("\n")
print("Marks column   :\n",da["Marks"])
print("\n")
print("Rows Marks >70 :\n",da[da["Marks"]>70])
```
